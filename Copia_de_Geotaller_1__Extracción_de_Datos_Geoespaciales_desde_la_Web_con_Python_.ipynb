##2. Extracción de Datos Espaciales con OpenStreetMap API

1. Petición de datos de hospitales y cafeterías mediante requests a Overpass API.
2. Transformación de datos a un GeoDataFrame.

```python
import requests
import geopandas as gpd
import pandas as pd
from shapely.geometry import Point
```

```python
# Definir la consulta en Overpass API para obtener hospitales en Bogotá
overpass_url = "http://overpass-api.de/api/interpreter"
query_hospitales = """
[out:json];
area["name"="Bogotá"]->.searchArea;
(
  node["amenity"="hospital"](area.searchArea);
);
out center;
"""

# Hacer la petición a la API
response = requests.get(overpass_url, params={"data": query_hospitales})
data = response.json()
```

```python
print(data)
```

```python
#Visualicemos de una manera más organiada la información provista en data
import json
print(json.dumps(data, indent=2))
```

Qué hay en *data["elements"]*?

```python
data["elements"]
```

```python
# Extraer información y convertirla en DataFrame
elements = data["elements"]
hospitales = []
for element in elements:
    if "lat" in element and "lon" in element:
        hospitales.append({
            "name": element["tags"].get("name", "Desconocido"),
            "lat": element["lat"],
            "lon": element["lon"]
        })
```

```python
print(hospitales)
```

```python
# Convertir a GeoDataFrame
df_hospitales = pd.DataFrame(hospitales)
print(df_hospitales)
```

```python
gdf_hospitales = gpd.GeoDataFrame(df_hospitales,
                                  geometry=gpd.points_from_xy(df_hospitales.lon, df_hospitales.lat),
                                  crs="EPSG:4326")
```

```python
# Mostrar los primeros registros
gdf_hospitales.tail()
```

**Qué hemos hecho hasta aquí?**:

* Se usa Overpass API para extraer hospitales en Bogotá.
* Se transforma la respuesta JSON en un GeoDataFrame con coordenadas.

##3. Visualización de Datos en un Mapa Interactivo

1. Uso de Folium para visualizar hospitales en un mapa interactivo.
2. Creación de capas de diferentes categorías (hospitales vs. cafeterías).

```python
import folium
```

```python
# Crear mapa base centrado en Bogotá
m = folium.Map(location=[4.7110, -74.0721], zoom_start=12)
m
```

```python
# Agregar hospitales al mapa
for idx, row in gdf_hospitales.iterrows():
    folium.Marker(
        location=[row["lat"], row["lon"]],
        popup=row["name"],
        icon=folium.Icon(color="red", icon="plus-sign"),
    ).add_to(m)

# Mostrar mapa
m
```

 Explicación del Código:

1. Se usa Folium para visualizar los hospitales en Bogotá.
2. Se agregan marcadores rojos (icon="plus-sign") en la ubicación de cada hospital.

##4. Integración de Múltiples Capas: Hospitales y Cafeterías (10 min)

* Se repite el proceso anterior para cafeterías.
*  Se combinan ambos datasets en un solo mapa.

```python
# Definir la consulta para obtener cafeterías en Bogotá
query_cafeterias = """
[out:json];
area["name"="Bogotá"]->.searchArea;
(
  node["amenity"="cafe"](area.searchArea);
);
out center;
"""

# Hacer la petición y extraer datos
response_cafes = requests.get(overpass_url, params={"data": query_cafeterias})
response_cafes
```

```python
data_cafes = response_cafes.json()
print(data_cafes)
```

```python
#Imprimimos en pantalla el JSON data_cafes de forma que se vea más legible
import json
print(json.dumps(data_cafes, indent=2))
```

```python
data_cafes["elements"]
```

```python
elements_cafes = data_cafes["elements"]
elements_cafes
```

```python
cafeterias = []
for element in elements_cafes:
    if "lat" in element and "lon" in element:
        cafeterias.append({
            "name": element["tags"].get("name", "Desconocido"),
            "lat": element["lat"],
            "lon": element["lon"]
        })

print(cafeterias)
```

```python
# Convertir a GeoDataFrame
df_cafeterias = pd.DataFrame(cafeterias)
df_cafeterias.tail()
```

```python
gdf_cafeterias = gpd.GeoDataFrame(df_cafeterias,
                                  geometry=gpd.points_from_xy(df_cafeterias.lon, df_cafeterias.lat),
                                  crs="EPSG:4326")
gdf_cafeterias.tail()
```

```python
# Integrar los datos en un mismo mapa

# Agregar cafeterías (azul)
for idx, row in gdf_cafeterias.iterrows():
    folium.Marker(
        location=[row["lat"], row["lon"]],
        popup=row["name"],
        icon=folium.Icon(color="blue", icon="coffee"),
    ).add_to(m)

# Mostrar mapa
m
```

 Explicación del Código:

1. Extraemos las cafeterías de la misma forma en la que obtuvimos la información relacionada a los hospitales.
2. Se agregan marcadores en el mapa con diferentes colores:

🔴 Hospitales → rojo
🔵 Cafeterías → azul

```python
#Descarga esta visualización en un archivo de formato .html
m.save("mapa_bogota.html")
from google.colab import files
files.download("mapa_bogota.html")
```
