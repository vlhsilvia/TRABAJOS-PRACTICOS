Podemos observar que la mayor parte de los geodataFrames tiene la misma estructura, a excepción de:

- transito_2019
- transiti_2020
- transito_2021

```python
# Creamos una lista con todos los GeoDataFrames
lista_gdfs = [transito_2014, transito_2015, transito_2016, transito_2017, transito_2018, transito_2019, transito_2020, transito_2021]

# Iteramos sobre la lista de GeoDataFrames para encontrar las columnas en común
columnas_comunes = set(lista_gdfs[0].columns)
for gdf in lista_gdfs[1:]:
    columnas_comunes = columnas_comunes.intersection(set(gdf.columns))
```

```python
# Imprimimos las columnas en común
print("Columnas en común en todos los GeoDataFrames:")
columnas_comunes
```

Veamos cuántas son:

```python
len(columnas_comunes)
```

Ahora lo que haremos ser[a lo siguiente:

1. Habiendo definido previamente cuáles son las columnas en común entre todos los geodataFrames ...
2. Vamos a *redefinir* cada uno de estos de forma en que coincidan exactamente entre sus columnas

```python
# Creamos una lista vacía para almacenar los GeoDataFrames procesados
gdfs_procesados = []

# Iteramos sobre la lista de GeoDataFrames original
for gdf in lista_gdfs:
  # Seleccionamos solo las columnas comunes
  gdf_procesado = gdf[list(columnas_comunes)]
  # Agregamos el GeoDataFrame procesado a la lista
  gdfs_procesados.append(gdf_procesado)
```

```python
gdfs_procesados #Es una lista de geodataFrames
```

Concatenamos los geodataFrames:

[Documentación del método *concat()*](https://pandas.pydata.org/docs/reference/api/pandas.concat.html)

```python
#Concatenamos los GeoDataFrames procesados
transito_concatenado = gpd.GeoDataFrame(pd.concat(gdfs_procesados, ignore_index=True), crs=transito_2014.crs)
#Resetea el index
transito_concatenado = transito_concatenado.reset_index(drop=True)
```

1️⃣ Preparación de los datos

```python
# Filtrar solo registros con coordenadas válidas
transito_concatenado = transito_concatenado.dropna(subset=['geometry'])
# Reseteamos el índice
transito_concatenado = transito_concatenado.reset_index(drop=True)
```

Adicionalmente queremos implementar un algoritmo de clustering: DB-SCAN para poder hacer una segmentación de las zonas según las particularidades de los accidentes de tránsito. Pero antes, debido a la gran cantidad de datos- y a las limitaciones de cómputo que tenemos, a pesar de estar haciendo uso de una instancia con GPU- vamos a aplicar el algoritmo usando sólo el 50% de los datos

### DB-SCAN

Si bien queremos analizar todos los datos, podemos reducir la cantidad de puntos usados en DBSCAN aplicando el algoritmo solo a una muestra representativa:

[Explicación de DB-SCAN](https://medium.com/@satpatishrimanta/clustering-by-dbscan-density-based-spatial-clustering-of-applications-with-noise-clearly-f93c5c72f706)

[Casos de Uso DB-SCAN:](https://medium.com/towards-data-science/identifying-topical-hot-spots-in-urban-areas-3c47cde5ae10)

```python
from sklearn.cluster import DBSCAN
from shapely.geometry import MultiPoint, Polygon
import numpy as np
```

```python
# Reducir la muestra al 50% de los datos (o menos si sigue dando error)
transito_sample = transito_concatenado.sample(frac=0.5, random_state=42)

# Extraer coordenadas para clustering
coords = np.array(list(zip(transito_sample.geometry.x, transito_sample.geometry.y)))
```

2️⃣ Paso 2(Adicional): Generación del Heatmap (Mapa de Calor)
Objetivo: Visualizar de manera exploratoria las zonas con mayor concentración de accidentes.

[Generando un heatmap en python con Folium](https://python-visualization.github.io/folium/latest/user_guide/plugins/heatmap.html)

```python
from folium import plugins

transito_sample_4326 = transito_sample.to_crs(epsg=4326)

# Crear mapa centrado en Medellín
map = folium.Map(location=[6.2442, -75.5812],
                 tiles='CartoDB positron',
                 zoom_start=13)

# Preparar datos para el heatmap
heat_data = [[point.y, point.x] for point in transito_sample_4326.geometry]

# Agregar capa de calor
plugins.HeatMap(heat_data).add_to(map)

# Mostrar mapa
map
```

3️⃣ Paso 3: Identificación de Zonas de Riesgo con DBSCAN
Objetivo: Agrupar accidentes en clusters densos y delimitar las zonas de mayor riesgo.

```python
# Aplicar DBSCAN para encontrar clusters de accidentes
dbscan = DBSCAN(eps=300, min_samples=50) #
transito_sample['cluster'] = dbscan.fit_predict(coords)
```

Cuántos clusters obtuvimos con el uso de este algoritmo?

```python

transito_sample['cluster'].unique()
```

2️⃣ Visualización de Clusters Identificados (Ajustado)

Ahora que hemos aplicado DBSCAN a una muestra de datos, ajustaremos la visualización para evitar ruido en los puntos y hacer que los clusters sean más visibles.

```python
fig, ax3 = plt.subplots(figsize=(12, 10))

# Graficar los puntos de accidentes sin cluster (outliers en negro)
transito_sample[transito_sample['cluster'] == -1].plot(
    ax=ax3, color='black', alpha=0.3, markersize=3, label="Outliers (Accidentes aislados)")

# Graficar los accidentes agrupados en clusters con colores distintos
for cluster_id in set(transito_sample['cluster']):
    if cluster_id != -1:
        transito_sample[transito_sample['cluster'] == cluster_id].plot(
            ax=ax3, markersize=8, alpha=0.7, label=f"Cluster {cluster_id}"
        )

# Agregar mapa base de OpenStreetMap
cx.add_basemap(ax3, source=cx.providers.OpenStreetMap.Mapnik, crs = transito_sample.crs)

# Ajustar títulos y etiquetas
ax3.set_title("Clusters de Accidentes de Tránsito en Medellín (Muestra 50%)")
#Agrega leyenda en la esquina superior derecha
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Crea un DataFrame con la información de cada uno de los clusters
cluster_info = transito_sample.groupby('cluster').size().reset_index(name='accident_count')

# Muestra el DataFrame
cluster_info
```

✅ Ajustes realizados:

1. Se redujo el tamaño de los puntos outliers para que no oculten clusters.
2. Se aumentó el tamaño de los puntos dentro de clusters para mejorar la visibilidad.
3. Se agregó una leyenda más clara para que los clusters sean interpretables.

3️⃣ Delimitación de Zonas de Riesgo con Convex Hulls (Ajustado)

Dado que trabajamos con una muestra, los clusters pueden ser más dispersos. Para evitar errores en el convex hull, agregamos un filtro para asegurarnos de que haya suficientes puntos en cada cluster.

```python
from shapely.geometry import MultiPoint

# Lista para almacenar las zonas de riesgo
zonas_riesgo = []

# Iteramos sobre cada cluster identificado
for cluster_id in set(transito_sample['cluster']):
    if cluster_id != -1:  # Omitimos los outliers
        puntos_cluster = transito_sample[transito_sample['cluster'] == cluster_id].geometry
        if len(puntos_cluster) > 5:  # Filtramos clusters demasiado pequeños
            hull = MultiPoint(puntos_cluster.tolist()).convex_hull  # Crear polígono envolvente
            zonas_riesgo.append({'cluster': cluster_id, 'geometry': hull})
```

```python
zonas_riesgo
```

```python
# Convertir en GeoDataFrame
zonas_riesgo_gdf = gpd.GeoDataFrame(zonas_riesgo, crs=transito_sample.crs)

zonas_riesgo_gdf.tail()
```

```python
zonas_riesgo_gdf.plot()
```

```python
# Graficar mapa con las zonas de riesgo delimitadas
fig, ax4 = plt.subplots(figsize=(12, 10))

# Graficar zonas de riesgo con polígonos envolventes
zonas_riesgo_gdf.plot(ax=ax4, edgecolor='red', facecolor='none', linewidth=2, label="Zonas de Riesgo")

# Agregar mapa base
cx.add_basemap(ax4, source=cx.providers.OpenStreetMap.Mapnik)

# Ajustar título y leyenda
plt.title("Zonas de Riesgo de Accidentes en Medellín (Clusters de Alta Densidad)")
plt.legend()
plt.show()
```

✅ Ajustes realizados:

Se descartaron clusters con menos de 5 puntos para evitar polígonos no representativos.
Los convex hulls reflejan mejor la distribución de accidentes en la muestra.
Mejoramos la visualización con bordes rojos y fondo de OpenStreetMap.
